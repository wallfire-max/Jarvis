## 3. Basic Inference (Zero-Shot)
Let's make a simple text completion request using `gemini-1.5-flash`.

```python
model = genai.GenerativeModel('gemini-1.5-flash')
response = model.generate_content("Explain quantum superposition in 2 sentences.")
print(response.text)
```

## 4. Few-Shot Prompting (Structured Output)
By providing examples before asking a question, we can guide the model to output a specific format.

```python
prompt = """
Task: Extract chemistry formulas.

Text: "Water consists of two hydrogen atoms bonded to an oxygen atom."
Formula: H2O
-------
Text: "Carbon dioxide is an acidic colorless gas with a density about 53% higher than that of dry air."
Formula: CO2
-------
Text: "Methane is a chemical compound with the chemical formula CH4. It is a group-14 hydride and the simplest alkane."
Formula: 
"""

response = model.generate_content(prompt)
print(f"Output: {response.text.strip()}")
```

## 5. System Instructions (Adding Personality)
We can configure a custom persona, such as a sarcastic assistant, using `system_instruction`.

```python
savage_model = genai.GenerativeModel(
    model_name='gemini-1.5-flash',
    system_instruction="You are a sarcastic, witty AI butler. Mock the user slightly, then answer their question."
)

response = savage_model.generate_content("What is the capital of France?")
print(response.text)
```

## 2. Listing Downloaded Models
Let's see what models you have downloaded on your system. Make sure the Ollama application is running in the background.

```python
import ollama

try:
    res = ollama.list()
    print("Downloaded models on your machine:")
    for model in res.get('models', []):
        print(f" - Name: {model['name']} | Size: {model['size'] / (1024**3):.2f} GB")
except Exception as e:
    print(f"Error connecting to Ollama: {e}. Is the Ollama app running?")
```

## 3. Basic Local Generation
We will call `llama3.2` locally on your machine.

```python
try:
    response = ollama.chat(
        model='llama3.2',
        messages=[{'role': 'user', 'content': 'List 3 elements of the periodic table.'}]
    )
    print(response['message']['content'])
except Exception as e:
    print(f"Failed to run chat: {e}")
```

## 4. Streaming Responses (Real-time Token Generation)
Instead of waiting for the model to finish its entire output, we can stream token by token in real-time.

```python
try:
    stream = ollama.chat(
        model='llama3.2',
        messages=[{'role': 'user', 'content': 'Write a 1-paragraph story about a robot.'}],
        stream=True,
    )

    for chunk in stream:
        print(chunk['message']['content'], end='', flush=True)
except Exception as e:
    print(f"Streaming failed: {e}")
```
